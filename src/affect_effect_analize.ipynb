```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

path = "../data/outlier_deal_for_nan_exp1.csv"

df = pd.read_csv(path)
```

```python
alertness_distribution = (
    df
    .groupby("Message Group")["Alertness Level"]
    .value_counts()
    .rename("count")
    .reset_index()
)

affect_distribution = (
    df
    .groupby("Message Group")["Affect Level"]
    .value_counts()
    .rename("count")
    .reset_index()
)
```

```python
print("Alertness:")
print(alertness_distribution.shape)

print("Affect:")
print(affect_distribution.shape)
```

```python
#9500 9600
#9465+35
print(
    alertness_distribution.groupby("Message Group")["count"].sum()
)

print(
    affect_distribution.groupby("Message Group")["count"].sum()
)
```

```python
import matplotlib.pyplot as plt

groups = ["neutral", "negative", "positive"]
colors = {
    "neutral": "green",
    "positive": "red",
    "negative": "blue"
}

plt.figure(figsize=(10, 6))

for group in groups:
    values = df.loc[
        df["Message Group"] == group,
        "Alertness Level"
    ].dropna()

    plt.hist(
        values,
        bins=20,
        alpha=0.5,
        label=group,
        color=colors[group]
    )

plt.xlabel("Alertness Level")
plt.ylabel("Count")
plt.title("Alertness Level distribution by Message Group")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

for group in groups:
    values = df.loc[
        df["Message Group"] == group,
        "Affect Level"
    ].dropna()

    plt.hist(
        values,
        bins=20,
        alpha=0.5,
        label=group,
        color=colors[group]
    )

plt.xlabel("Affect Level")
plt.ylabel("Count")
plt.title("Affect Level distribution by Message Group")
plt.legend()

plt.tight_layout()
plt.show()
```

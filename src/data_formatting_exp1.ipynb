```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

path = "../data/exp1_824.csv"
df = pd.read_csv(path)
```

```python
# 削除したい列名
drop_cols = [
    "StartDate",
    "EndDate",
    "Status",
    "IPAddress",
    "Progress",
    "Duration (in seconds)",
    "Finished",
    "RecordedDate",
    "RecipientLastName",
    "RecipientFirstName",
    "RecipientEmail",
    "ExternalReference",
    "LocationLatitude",
    "LocationLongitude",
    "DistributionChannel",
    "UserLanguage",
    "最後に表示されたフロー要素ID",
    "最後に表示された質問ID",
    "D1",
    "D2_1",
    "B1"
]

# 不要列を削除してdfを更新
df = df.drop(columns=drop_cols)

print(df.shape)
```

```python
drop_cols = df.columns[
    df.columns.astype(str).str.contains(
        "First Click|Last Click|Click Count|B2",
        na=False
    )
]

df = df.drop(columns=drop_cols)
```

```python
# 全ての行がNaNの列を削除

data_body = df

# 全てNaNの列を判定
cols_to_drop = data_body.columns[data_body.isna().all()]

# 削除する列名を表示
print("削除する列数:", len(cols_to_drop))
print("削除する列名:")

for col in cols_to_drop:
    print(col)

# 列を削除a
df = data_body.drop(columns=cols_to_drop)

print("残った列数:", len(df.columns))
print("残った行数:", len(df))
```

```python
df.shape
```

```python
n_frame_blocks = 20
affect_per_frame = 5
affect_trial_size = 3

start_candidates = [1, 341, 641]

records = []

for row_idx in range(len(df)):

    row = df.iloc[row_idx]

    # ==========================================
    # ① Affectの開始位置を決定
    # ==========================================

    if pd.notna(row.iloc[1]):
        start_col = 1
        condition = 1

    elif pd.notna(row.iloc[341]):
        start_col = 341
        condition = 2

    elif pd.notna(row.iloc[641]):
        start_col = 641
        condition = 3

    else:
        raise ValueError(
            f"CellNotFound: row={row_idx}, "
            f"columns={start_candidates}"
        )

    # ==========================================
    # ② Frame Choiceを最初の領域から探索
    # ==========================================

    frame_values = []

    # Frame Choiceはconditionに関係なく
    # 1列目から探索
    for col_idx in range(1, len(row)):

        value = row.iloc[col_idx]

        if "半分" in str(value):

            frame_values.append([
                row.iloc[col_idx - 1],
                value
            ])

            if len(frame_values) == n_frame_blocks:
                break

    # ==========================================
    # Frame数を確認
    # ==========================================

    if len(frame_values) != n_frame_blocks:
        raise ValueError(
            f"Frame Choice不足: "
            f"row={row_idx}, "
            f"condition={condition}, "
            f"取得数={len(frame_values)}"
        )

    # ==========================================
    # ③ Frame 20ブロック
    # ==========================================

    for i in range(n_frame_blocks):

        frame_vals = frame_values[i]

        # ======================================
        # ④ 1 FrameにつきAffect 5試行
        # ======================================

        for j in range(affect_per_frame):

            affect_idx = i * affect_per_frame + j

            # ----------------------------------
            # Affectの開始列
            # ----------------------------------

            if condition == 1:

                # 15列(Affect) + 2列(Frame)
                affect_start = (
                    start_col
                    + i * 17
                    + j * 3
                )

            else:

                # 条件2・3
                affect_start = (
                    start_col
                    + affect_idx * 3
                )

            # ----------------------------------
            # Affect 3列
            # ----------------------------------

            affect_vals = row.iloc[
                affect_start:affect_start + affect_trial_size
            ].values

            # ----------------------------------
            # 1レコード
            # ----------------------------------

            record = {
                "responseId": row.iloc[0],

                "frame_block": i,
                "affect_trial": j,

                "Page Submit_al": affect_vals[0],
                "Alertness Level": affect_vals[1],
                "Affect Level": affect_vals[2],

                "frame_1": frame_vals[0],
                "frame_2": frame_vals[1],

                "Message Group": row["__js_MessageGroup"]
            }

            records.append(record)


new_df = pd.DataFrame(records)

print("Shape:", new_df.shape)
```

```python
new_df["frame_2"] = new_df["frame_2"].map({
    "半分も入っている": 0,
    "半分しか入っていない": 1
})
```

```python
new_df = new_df.rename(columns={
    "affect_trial":"ref_subblock",
    "frame_1":"Page Submit_fc",
    "frame_2":"frame_choice"
})
```

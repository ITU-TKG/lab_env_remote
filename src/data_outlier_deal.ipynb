```python
import pandas as pd
import numpy as np

path = "../data/formatted_trial_flagged_exp2.csv"
df = pd.read_csv(path)
```

```python
#実験1
colum_name = "is_outlier"
targets = [
    "Affect Level",
    "Alertness Level",
    "Page Submit_al",
    "Page Submit_fc",
    "frame_choice"
]
n_outliers = df[colum_name].sum()
df.loc[df[colum_name], targets] = np.nan
print(f"NaNにした行数: {n_outliers}")
```

```python
#実験2
colum_name = "is_outlier"
targets = [
    "Affect Level",
    "Alertness Level",
    "Page Submit_al",
    "Water Level",
    "Page Submit_wl",
    "Page Submit_fc",
    "frame_choice"
]

n_outliers = df[colum_name].sum()
df.loc[df[colum_name], targets] = np.nan
print(f"NaNにした行数: {n_outliers}")
```

```python
#実験1
df.to_csv("../data/outlier_deal_for_nan_exp1.csv") 
```

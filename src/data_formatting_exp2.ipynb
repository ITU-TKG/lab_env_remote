```python
path = "../data/exp2_824.csv" #適宜変更
```

```python
import pandas as pd

data = pd.read_csv(path)
print(data.shape)
df = data
```

```python
ip_data = df["IPAddress"]

#重複IPの検知

ip_counts = df["IPAddress"].value_counts()

print(ip_counts[ip_counts > 1])
```

```python
# 削除したい列名
drop_cols = [
    "StartDate",
    "EndDate",
    "Status",
    "IPAddress",
    "Progress",
    "Duration (in seconds)",
    "Finished",
    "RecordedDate",
    "RecipientLastName",
    "RecipientFirstName",
    "RecipientEmail",
    "ExternalReference",
    "LocationLatitude",
    "LocationLongitude",
    "DistributionChannel",
    "UserLanguage",
    "最後に表示されたフロー要素ID",
    "最後に表示された質問ID",
    "D1",
    "D2_1",
    "B1"
]

# 不要列を削除してdfを更新
df = df.drop(columns=drop_cols)

print(df.shape)
```

```python
drop_cols = df.columns[
    df.columns.astype(str).str.contains(
        "First Click|Last Click|Click Count|B2",
        na=False
    )
]

df = df.drop(columns=drop_cols)
```

```python
df.shape
```

```python
# 全ての行がNaNの列を削除

data_body = df

# 全てNaNの列を判定
cols_to_drop = data_body.columns[data_body.isna().all()]

# 削除する列名を表示
print("削除する列数:", len(cols_to_drop))
print("削除する列名:")

for col in cols_to_drop:
    print(col)

# 列を削除a
df = data_body.drop(columns=cols_to_drop)

print("残った列数:", len(df.columns))
print("残った行数:", len(df))
```

```python
responseId = df.iloc[:,0]
frame_choice_answer = df.iloc[:,1:41]
reference_answer = df.iloc[:,41:]
```

```python
n_frame_blocks = 20
frame_block_size = 2

n_ref_subblocks = 5
ref_subblock_size = 5

records = []
message_group = df["__js_MessageGroup"]

for row_idx in range(len(df)):

    resp_id = responseId.iloc[row_idx]

    # この行のReferenceを取得
    ref_row = reference_answer.iloc[row_idx]

    # NaNではない値だけを取り出す
    ref_values = ref_row.dropna().values

    # Frameは従来通り20ブロック
    for i in range(n_frame_blocks):

        # --------------------
        # Frame：2列
        # --------------------
        f_start = i * frame_block_size

        frame_vals = frame_choice_answer.iloc[
            row_idx,
            f_start:f_start + frame_block_size
        ].values

        # --------------------
        # Reference：5列 × 5回
        # --------------------
        for j in range(n_ref_subblocks):

            # このFrameに対応するReferenceの位置
            ref_start = (i * n_ref_subblocks + j) * ref_subblock_size

            ref_vals = ref_values[
                ref_start:ref_start + ref_subblock_size
            ]

            # --------------------
            # レコード作成
            # --------------------
            record = {
                "responseId": resp_id,
                "frame_block": i,
                "ref_subblock": j,
                "Message Group": message_group.iloc[row_idx],                
            }

            record.update({
                f"frame_{k+1}": v
                for k, v in enumerate(frame_vals)
            })

            record.update({
                f"ref_{k+1}": v
                for k, v in enumerate(ref_vals)
            })

            records.append(record)

new_df = pd.DataFrame(records)
```

```python
new_df["frame_2"] = new_df["frame_2"].map({
    "半分も入っている": 0,
    "半分しか入っていない": 1
})
```

```python
new_df = new_df.rename(columns={
    "ref_1": "Page Submit_wl",
    "ref_2": "Water Level",
    "ref_3": "Page Submit_al",
    "ref_4": "Alertness Level",
    "ref_5": "Affect Level",
    "frame_1":"Page Submit_fc",
    "frame_2":"frame_choice"
})
```

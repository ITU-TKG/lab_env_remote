```python
import matplotlib.pyplot as plt
import japanize_matplotlib

plt.hist(df["Page Submit_fc"].dropna(), bins=20, range=(0, 100))

plt.xlabel("Page Submit_fc")
plt.ylabel("個数")
plt.title("Page Submit_fc の分布")

plt.show()
```
